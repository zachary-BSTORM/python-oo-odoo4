# L'encapsulation FR

L'encapsulation permet de **contrôler l'accès** aux attributs d'une classe. On distingue trois niveaux :

| Niveau | Préfixe | Accès | Exemple |
|---|---|---|---|
| **Public** | aucun | Partout | `self.nom` |
| **Protégé** | `_` | Simple **convention** : « réservé à la classe et ses sous-classes » — Python ne bloque rien | `self._nom` |
| **Privé** | `__` | Classe uniquement (via *name mangling*, contournable) | `self.__nom` |

> ⚠️ Python ne fait **pas de vraie encapsulation**. Il utilise des **conventions de nommage**. Par défaut, tous les attributs sont publics.
>
> Le préfixe `_` (protégé) en particulier n'est **qu'un accord entre développeurs** : c'est un panneau « privé, merci de ne pas toucher ». Techniquement, `objet._nom` reste accessible partout — Python n'interdit rien, il compte sur la discipline du lecteur.

### Le *name mangling* (privé)

Un attribut préfixé par `__` est renommé automatiquement par Python pour le rendre moins accessible :
- `self.__name` dans la classe `Chien` → devient `_Chien__name` à l'exécution

# Encapsulation EN

Encapsulation lets you **control access** to a class's attributes. There are three levels:

| Level | Prefix | Access | Example |
|---|---|---|---|
| **Public** | none | Everywhere | `self.nom` |
| **Protected** | `_` | Just a **convention**: "reserved for the class and its subclasses" — Python blocks nothing | `self._nom` |
| **Private** | `__` | Class only (via *name mangling*, can be bypassed) | `self.__nom` |

> ⚠️ Python does **not have real encapsulation**. It relies on **naming conventions**. By default, all attributes are public.
>
> The `_` (protected) prefix in particular is **only an agreement between developers**: it's a "private, please don't touch" sign. Technically, `objet._nom` remains accessible everywhere — Python forbids nothing, it relies on the reader's discipline.

### *Name mangling* (private)

An attribute prefixed with `__` is automatically renamed by Python to make it less accessible:
- `self.__name` in the `Chien` class → becomes `_Chien__name` at runtime

```python
class Person:
    def __init__(self,name : str,email : str,password:str,age:int):
        self.name = name
        self._email = email
        self.__password = password
        self.age = age
        
p_1 = Person("John","JohnDoe@mail.com","test1234",30)

print(p_1.name)
print(p_1._email)
print(p_1._Person__password) # ici accès à l'attribut privé
# here we access the private attribute

p_1._Person__password = "test" # modifiable 
# can be modified
print(p_1._Person__password)
```

# Getter & setter

```python
class Person:
    def __init__(self,name : str,email : str,password:str,age:int):
        self.name = name
        self._email = email
        self.__password = password # création si n'existe pas
        # created if it doesn't exist
        self.age = age
    
        
    def get_password(self):
        return self.__password
    
    def set_password(self,value):
        self.__password = value
        
p_1 = Person("John","JohnDoe@mail.com",30)

# setter
p_1.set_password("admin1234")

# getter
print(p_1.get_password())
```

# Décorateur **@property**
# **@property** decorator

| Décorateur | Rôle | Déclencheur |
|---|---|---|
| `@property` | Getter | `obj.attr` |
| `@attr.setter` | Setter | `obj.attr = valeur` |
| `@attr.deleter` | Deleter | `del obj.attr` |

| Decorator | Role | Trigger |
|---|---|---|
| `@property` | Getter | `obj.attr` |
| `@attr.setter` | Setter | `obj.attr = value` |
| `@attr.deleter` | Deleter | `del obj.attr` |

```python
class Computer:
    def __init__(self,brand):
        self.__brand = brand
        
    # accès à l'attribut pour lecture
    # read access to the attribute
    @property
    def brand(self):
        return self.__brand
    
    # modification de l'attribut
    # modifying the attribute
    @brand.setter
    def brand(self,value):
        if len(value) < 3:
            raise Exception("La marque doit faire au moins 3 caractères")
        else:
            self.__brand = value
    
    # suppression de l'attribut      
    # deleting the attribute
    @brand.deleter
    def brand(self):
        del self.__brand
```
