```python
class Car:
    def __init__(self):
        self.marque = "honda"
        self.model = "civic"
        self.nb_seat = 5
        
    # représentation défini pour l'objet 
    def __repr__(self):
        return f"Le vehicule de marque {self.marque} - model {self.model}"

    # Représentation de l'objet en str ( str() - print() )
    def __str__(self):
        return f"Affichage des données : {self.marque} - {self.model}"
    
    # modifie le comportement de len() 
    def __len__(self):
        return self.nb_seat
    
    # crée un accès par clé 
    def __getitem__(self,value):
        match value:
            case "marque":
                return self.marque
            case "model":
                return self.model
            case "nb_seat":
                return self.nb_seat
            
new_car = Car()

print(repr(new_car))

print(str(new_car))
print(new_car)

print(len(new_car))

print(new_car["marque"])
```

| Méthode | Opérateur | Description |
|---|---|---|
| `__eq__` | `==` | Égalité |
| `__ne__` | `!=` | Différence |
| `__lt__` | `<` | Inférieur à |
| `__le__` | `<=` | Inférieur ou égal |
| `__gt__` | `>` | Supérieur à |
| `__ge__` | `>=` | Supérieur ou égal |

```python
class Car(object):
    def __init__(self,nb_seat):
        self.marque = "honda"
        self.model = "civic"
        self.nb_seat = nb_seat
        
        
    # ==
    def __eq__(self,other):
        return self.nb_seat == other.nb_seat
    
    # !=
    def __ne__(self,other):
        return self.marque != other.marque
    
    # <
    def __lt__(self,other):
        return self.nb_seat < other.nb_seat
        
        
    # <=
    def __le__(self,other):
        return self.nb_seat <= other.nb_seat
    
    # >
    def __gt__(self,other):
        return self.nb_seat != other.nb_seat
    # >=
    def __ge__(self,other):
        return self.nb_seat >= other.nb_seat
    
if __name__ == "__main__":

    new_car = Car(5)
    new_car_2 = Car(5)
    
    print("==")
    print(new_car == new_car_2)
    
    print("!=")
    print(new_car != new_car_2)
    
    print("<")
    print(new_car < new_car_2)
    
    print("<=")
    print(new_car <= new_car_2)
    
    print(">")
    print(new_car > new_car_2)
    
    print(">=")
    print(new_car >= new_car_2)
```

### Collections et itération
| Méthode | Déclenchée par | Description |
|---|---|---|
| `__len__(self)` | `len(obj)` | Longueur de l'objet |
| `__getitem__(self, key)` | `obj[key]` | Accès par index/clé |
| `__setitem__(self, key, val)` | `obj[key] = val` | Affectation par index/clé |
| `__delitem__(self, key)` | `del obj[key]` | Suppression par index/clé |
| `__contains__(self, item)` | `item in obj` | Test d'appartenance |
| `__iter__(self)` | `iter(obj)`, `for x in obj` | Retourne un itérateur |
| `__next__(self)` | `next(obj)` | Prochain élément de l'itérateur |
| `__reversed__(self)` | `reversed(obj)` | Itérateur inversé |
