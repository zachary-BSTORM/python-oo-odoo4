# Création de classe
- Convention de nommage : PascalCase

# Class creation
- Naming convention: PascalCase

```python
# Création d'une classe vide 
# Creation of an empty class
class Empty:
    pass

# Création d'une classe avec constructeur
# Creation of a class with a constructor
class Animal:
    # constructeur ( init)
    # constructor ( init)
    # self : référence à l'instance courante
    # self : reference to the current instance
    def __init__(self,name : str):
        self.name = name
```

# Ajout et suppression d'attributs
# Adding and removing attributes

```python
chat = Animal()

# Ici l'attribut new_info sera crée
# Here the new_info attribute will be created
chat.new_info = "nouvelle information"

# Suppression d'un attribut
# Removing an attribute
del chat.new_info
```

# Methods

```python
# premier paramètre l'instance elle-même
# first parameter is the instance itself
def make_noise(self):
    print(f"l'animal {self.name} fait du bruit")
```

# Class documentation

- visible when hovering over the class
- accessible by printing __doc__ : print(Animal.__doc__)

```python
class Animal:
    """
    représente un animal 
    represents an animal
    name : str
    color : str
    age : int
    """

print(Animal.__doc__)
```

# Attributs de classe
# Class attributes

```python
class Car:
    # lié à la classe
    # bound to the class
    nb_roues = 4
    
    # lié à l'instance
    # bound to the instance
    def __init__(self,nom : str):
        self.nom = nom
        
new_car = Car("toyota")


print(Car.nb_roues) # 4
print(new_car.nb_roues) # 4

# ici on crée un nouvel attribut nb_roues
# here we create a new nb_roues attribute
new_car.nb_roues = 5

print(new_car.nb_roues) # 5
print(Car.nb_roues) # 4
```
