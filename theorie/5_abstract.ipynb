# Les classes abstraites
# Abstract classes

- Une classe abstraite permet de définir qu'un ou plusieurs éléments d'une classe doivent être implémentés dans les classes qui en héritent
- An abstract class lets you declare that one or more members of a class must be implemented by the classes that inherit from it

<br>

- Elle sert de **contrat** et ne peut pas être instanciée directement
- It acts as a **contract** and cannot be instantiated directly

<br>

- Elle peut contenir des **méthodes abstraites** (sans code) et des **méthodes concrètes** (avec code)
- It can contain **abstract methods** (no code) and **concrete methods** (with code)

## Le module `abc`
## The `abc` module

- **`ABC`** : la classe de base dont on hérite pour rendre une classe abstraite
- **`ABC`**: the base class you inherit from to make a class abstract

<br>

- **`@abstractmethod`** : le décorateur qui marque une méthode comme obligatoire dans les classes enfants
- **`@abstractmethod`**: the decorator that marks a method as mandatory in the child classes

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):

    @abstractmethod
    def mouvement(self):
        pass


# 1) Une classe abstraite ne peut pas être instanciée directement
# 1) An abstract class cannot be instantiated directly
try:
    vehicule_1 = Vehicule()
except TypeError as e:
    print("TypeError :", e)


# 2) Une classe enfant qui n'implémente pas tout le contrat reste abstraite elle aussi
# 2) A child class that does not implement the whole contract stays abstract too
class Trottinette(Vehicule):
    pass  # mouvement() n'est pas implémentée / mouvement() is not implemented

try:
    trottinette_1 = Trottinette()
except TypeError as e:
    print("TypeError :", e)


# 3) Une classe enfant qui implémente tout le contrat s'instancie normalement
# 3) A child class that implements the whole contract can be instantiated normally
class Velo(Vehicule):
    def mouvement(self):
        return "Le vélo pédale / the bike pedals"

velo_1 = Velo()
print(velo_1.mouvement())
```

## Propriété abstraite
## Abstract property

- Pour rendre une propriété abstraite, l'ordre des décorateurs est obligatoire : `@property` **au-dessus** de `@abstractmethod`
- To make a property abstract, the decorator order is mandatory: `@property` **above** `@abstractmethod`

<br>

- La classe enfant l'implémente ensuite avec un simple `@property`
- The child class then implements it with a plain `@property`

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):

    @property
    @abstractmethod
    def serial_number(self):  # propriété abstraite : @property PUIS @abstractmethod
        pass                  # abstract property: @property THEN @abstractmethod

    @abstractmethod
    def show_infos(self):
        pass

class Moto(Vehicule):

   @property
   def serial_number(self): # la classe Moto hérite de la classe abstraite Vehicule et doit donc implémenter la propriété serial_number
       # the Moto class inherits from the abstract class Vehicule and therefore must implement the serial_number property
       return "bdfqs4d56sfd465sd from moto"

   def show_infos(self): # ... et la méthode show_infos, sinon Moto reste abstraite à son tour
       # ... and the show_infos method, otherwise Moto stays abstract too
       return "informations from moto"


# vehicule_1 = Vehicule() # erreur : la classe Vehicule est abstraite et ne peut pas être instanciée
# error: the Vehicule class is abstract and cannot be instantiated
moto_1 = Moto()

print(moto_1.show_infos())
print(moto_1.serial_number)   # une propriété se lit SANS parenthèses
# a property is read WITHOUT parentheses
```

## Le contrat en action : le polymorphisme
## The contract in action: polymorphism

- Toutes les classes enfants respectent le même contrat : on peut donc les traiter de la même façon, sans connaître leur type réel
- All child classes respect the same contract: we can therefore handle them the same way, without knowing their actual type

```python
class Voiture(Vehicule):

    @property
    def serial_number(self):
        return "ZX-4785-PL from car"

    def show_infos(self):
        return "informations from car"


# Le contrat garantit que CHAQUE véhicule possède show_infos() et serial_number
# The contract guarantees that EVERY vehicle has show_infos() and serial_number
parc = [Moto(), Voiture()]

for vehicule in parc:
    print(vehicule.show_infos(), "-", vehicule.serial_number)
```

## Méthodes concrètes dans une classe abstraite
## Concrete methods in an abstract class

- Une classe abstraite peut aussi contenir des méthodes **avec** du code : les enfants en héritent telles quelles
- An abstract class can also contain methods **with** code: the children inherit them as they are

<br>

- C'est le mélange des deux mondes : du code commun partagé + un contrat imposé
- It is the best of both worlds: shared common code + an enforced contract

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):

    def demarrer(self):                  # méthode CONCRÈTE : les enfants l'héritent telle quelle
        # CONCRETE method: the children inherit it as it is
        return "Le moteur démarre / the engine starts"

    @abstractmethod
    def mouvement(self):                 # méthode ABSTRAITE : chaque enfant DOIT l'implémenter
        # ABSTRACT method: every child MUST implement it
        pass


class VoitureElectrique(Vehicule):
    def mouvement(self):
        return "La voiture électrique roule / the electric car drives"


voiture_electrique_1 = VoitureElectrique()
print(voiture_electrique_1.demarrer())    # héritée de la classe abstraite / inherited from the abstract class
print(voiture_electrique_1.mouvement())   # imposée par le contrat / enforced by the contract
```

## Synthèse
## Summary

| À retenir / Key point | En une phrase / In one sentence |
|---|---|
| Classe abstraite / Abstract class | Un modèle qui impose un contrat, jamais instancié directement / A template that enforces a contract, never instantiated directly |
| `ABC` | La classe de base du module `abc` dont on hérite / The base class from the `abc` module you inherit from |
| `@abstractmethod` | Marque une méthode obligatoire dans les enfants / Marks a method as mandatory in the children |
| Enfant incomplet / Incomplete child | Reste abstrait → `TypeError` à l'instanciation / Stays abstract → `TypeError` on instantiation |
| Propriété abstraite / Abstract property | `@property` au-dessus de `@abstractmethod` / `@property` above `@abstractmethod` |
| Méthode concrète / Concrete method | Une classe abstraite peut aussi fournir du code hérité tel quel / An abstract class can also provide code inherited as is |

> **Question d'entretien** : « À quoi sert une classe abstraite ? » — À imposer un contrat commun à une famille de classes : chaque enfant doit implémenter les méthodes marquées `@abstractmethod`, sinon Python refuse l'instanciation avec une `TypeError`.
>
> **Interview question**: "What is an abstract class for?" — To enforce a shared contract on a family of classes: every child must implement the methods marked `@abstractmethod`, otherwise Python refuses the instantiation with a `TypeError`.
