# Héritage
# Inheritance

- Permet à une classe **Enfant** de récupérer le comportement d'une ou plusieurs classes **Parent**
- Allows a **Child** class to inherit the behavior of one or more **Parent** classes

<br>

- permet d'alléger le code et d'éviter la répétition de code 
- Makes code lighter and avoids code duplication

<br>

- facilite la maintenance et l'organisation du code 
- Makes code easier to maintain and organize

```python
class Vehicule:
    def __init__(self, marque, modele):
        # Attributs d'instance : chaque objet a ses propres valeurs
        # Instance attributes: each object has its own values
        self.marque = marque
        self.modele = modele

    def afficher_info(self):
        print(f"Marque: {self.marque}")
        print(f"Modèle: {self.modele}")


class Voiture(Vehicule):  # Voiture hérite de Vehicule
    # Voiture inherits from Vehicule
    pass                  # rien à ajouter : elle hérite du constructeur et des méthodes
    # nothing to add: it inherits the constructor and the methods


# Voiture n'a rien défini elle-même : le constructeur (__init__) et la méthode
# afficher_info viennent entièrement de la classe parente Vehicule
# Voiture defined nothing itself: the constructor (__init__) and the
# afficher_info method come entirely from the parent class Vehicule
ma_voiture = Voiture("Toyota", "Corolla")

print("Informations de la voiture :")
ma_voiture.afficher_info()  # Méthode héritée de Vehicule
# Method inherited from Vehicule
```

## La méthode `super()`

`super()` permet d'appeler une méthode de la **classe parente** depuis la sous-classe. C'est particulièrement utile pour :
- Appeler le constructeur parent
- Étendre une méthode héritée sans la remplacer entièrement

## The `super()` method

`super()` lets you call a method of the **parent class** from the subclass. It is especially useful to:
- Call the parent constructor
- Extend an inherited method without replacing it entirely

```python
class Voiture(Vehicule):
    def __init__(self, marque, modele, couleur):
        super().__init__(marque, modele)   # appelle le constructeur de Vehicule
        # calls the Vehicule constructor
        self.couleur = couleur             # puis ajoute son propre attribut
        # then adds its own attribute

    def afficher_info(self):
        super().afficher_info()            # étend la méthode héritée sans la remplacer
        # extends the inherited method without replacing it
        print(f"Couleur: {self.couleur}")


# Instanciation : on observe le constructeur parent + l'extension de la méthode
# Instantiation: we see the parent constructor + the method extension
ma_voiture = Voiture("Toyota", "Corolla", "Rouge")
ma_voiture.afficher_info()
```

# Héritage multiple
# Multiple inheritance

```python
class Animal:
    def __init__(self,name : str,color : str , age : int):
        self.name = name
        self.color = color
        self.age = age
        
    
    def make_noise(self):
        return "The animal make noise "
        
class Breed:
    def __init__(self,breed_name : str , price : int):
        self.breed_name = breed_name
        self.price = price
        
    def get_information(self):
        return f"Breed : {self.breed_name} - price : {self.price}"
    
    def make_noise(self):
        return "this breed is noisy "
    
class Chien(Animal,Breed):
    def __init__(self,toy : str,hungry : bool,breed_name,price,name,color,age):
        self.toy = toy
        self.hungry = hungry
        Breed.__init__(self,breed_name,price)
        Animal.__init__(self,name,color,age)
        

    def make_noise(self):
        return "The dog make noise" + Animal.make_noise(self) + Breed.make_noise(self)
    
rex = Chien("bones",True,"berger allemand",500,"rex","brun",5)
```

# **kwargs dans le **__init__**
# **kwargs in **__init__**

```python
class Animal:
    def __init__(self,name : str,color : str , age : int,**kwargs):
        self.name = name
        self.color = color
        self.age = age
        
    
    def make_noise(self):
        return "The animal make noise "
        
class Breed:
    def __init__(self,breed_name : str , price : int,**kwargs):
        self.breed_name = breed_name
        self.price = price
        
    def get_information(self):
        return f"Breed : {self.breed_name} - price : {self.price}"
    
    def make_noise(self):
        return "this breed is noisy "
    
class Chien(Animal,Breed):
    def __init__(self,toy : str,hungry : bool,**kwargs):
        self.toy = toy
        self.hungry = hungry
        Breed.__init__(self,**kwargs)
        Animal.__init__(self,**kwargs)
        

    def make_noise(self):
        return "The dog make noise" + Animal.make_noise(self) + Breed.make_noise(self)
    
rex = Chien("bones",True,name="rex",color="brun",age=5,breed_name="berger allemand",price=500)
```
