# Création de classe
- Convention de nommage : PascalCase

```python

class Animal:
    # constructeur ( init)
    # self : référence à l'instance courante
    def __init__(self,name : str):
        self.name = name
```

# Ajout et suppression d'attributs

```python
chat = Animal()

# Ici l'attribut new_info sera crée
chat.new_info = "nouvelle information"

# Suppression d'un attribut
del chat.new_info
```

# Methodes

```python
# premier paramètre l'instance elle-même
def make_noise(self):
    print(f"l'animal {self.name} fait du bruit")
```

# Documentation de la classe

- accessible lors du survol de la classe
- accessible en affichant __doc__ : print(Animal.__doc__)

```python
class Animal:
    """
    représente un animal 
    name : str
    color : str
    age : int
    """

print(Animal.__doc__)
```

# Attributs de classe

```python
class Car:
    # lié à la classe
    nb_roues = 4
    
    # lié à l'instance
    def __init__(self,nom : str):
        self.nom = nom
        
new_car = Car("toyota")


print(Car.nb_roues) # 4
print(new_car.nb_roues) # 4

# ici on crée un nouvel attribut nb_roues
new_car.nb_roues = 5

print(new_car.nb_roues) # 5
print(Car.nb_roues) # 4
```
