## Les algorithmes de tri

Pour les imaginer, on se représente une rangée d'enfants à classer par taille.

| Tri | Image | Complexité |
| --- | --- | --- |
| À bulles | On compare les voisins et on échange ; le plus grand remonte au bout comme une bulle | O(n²) |
| Par sélection | On cherche le plus petit et on le met devant, puis on recommence avec le reste | O(n²) |
| Par insertion | Comme ranger des cartes en main : on glisse chaque carte à sa place | O(n²), presque O(n) si déjà presque trié |
| Fusion | On coupe jusqu'aux éléments seuls, puis on recolle deux à deux dans l'ordre | O(n log n) |
| Quicksort | On choisit un pivot : petits à gauche, grands à droite, puis on recommence | O(n log n) en moyenne, O(n²) au pire |
| Timsort | La bibliothécaire qui garde les rangées déjà en ordre, range les petits bouts à la main, puis recolle | O(n log n), presque O(n) si déjà presque trié |

```python
# QuickSort :  rapide 
from random import choice,randint
import sys

ma_list = [randint(1, 1000) for _ in range(10000)]


def quicksort(liste):
    if len(liste) <= 1:
        return liste
    pivot = choice(liste)
    plus_petits = [x for x in liste if x < pivot]
    egaux       = [x for x in liste if x == pivot]
    plus_grands = [x for x in liste if x > pivot]
    return quicksort(plus_petits) + egaux + quicksort(plus_grands)

quicksort(ma_list)
```

```python
# tri fusion : rapide


def tri_fusion(liste):
    if len(liste) <= 1:
        return liste
    
    millieu = len(liste) // 2
    gauche = tri_fusion(liste[millieu:])
    droite = tri_fusion(liste[:millieu])
    
    result , i , j = [] , 0 , 0
    
    while i < len(gauche) and j < len(droite):
        if gauche[i] <= droite[j]:
            result.append(gauche[i])
            i += 1
        else :
            result.append(droite[j])
            j += 1
    return result + gauche[i:] + droite[j:]

ma_list = [randint(1, 1000) for _ in range(100)]

tri_fusion(ma_list)
```

```python
# tri à bulles : très lent
from random import randint
def tri_bulles(liste):
    n = len(liste)
    
    for i in range(n - 1):
        for j in range( n - 1 - i):
            if liste[j] < liste[j + 1]:
                liste[j],liste[j + 1] = liste[j + 1],liste[j]
    return liste

ma_list = [randint(1, 1000) for _ in range(10000)]

tri_bulles(ma_list)
```

```python
# tri selection : très lent
from random import randint

def tri_selection(liste):
    n = len(liste)
    
    for i in range(n - 1):
        index_min = 0
        
        for j in range(i + 1 , n):
            if liste[j] < liste[index_min]:
                index_min = j
        liste[i],liste[index_min] = liste[index_min],liste[i]
    return liste
        
ma_list = [randint(1, 1000) for _ in range(100)]

tri_selection(ma_list)
```

```python
from timeit import timeit

ma_list = [randint(1, 1000) for _ in range(1000)]

temps_1 = timeit(
    stmt=lambda : quicksort(ma_list),
    number=5000
)
print(f"quickSort : {temps_1}")


temps_1 = timeit(
    stmt=lambda : tri_fusion(ma_list),
    number=5000
)
print(f"tri fusion : {temps_1}")


temps_1 = timeit(
    stmt=lambda : tri_bulles(ma_list),
    number=5000
)
print(f"tri bulles : {temps_1}")

temps_1 = timeit(
    stmt=lambda : tri_selection(ma_list),
    number=5000
)
print(f"tri selection : {temps_1}")

temps_1 = timeit(
    stmt=lambda : sorted(ma_list),
    number=5000
)
print(f"sorted : {temps_1}")
```
