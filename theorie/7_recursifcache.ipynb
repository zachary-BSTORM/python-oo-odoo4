```python

from timeit import timeit # permet de calculer le temps d'execution
from functools import cache # permet de mettre en cache les éléments calculé ( memoisation )

def factoriel(n):
    resultat = 1
    for i in range(2 , n + 1):
        resultat *= i
    return resultat

def facto(n):
    result = 1
    i = 1
    while i <= n :
        result = result * i
        i += 1
        
    return result

@cache # utilise la memoisation 
def facto_recursif(n):
    if n <= 1:
        return 1
    return n * facto_recursif(n - 1)

# memoisation fait main
values = {}
def facto_without_cache(n):
    global values
    
    if n in values:
        return values[n]
    else:
        resultat = n * facto_without_cache( n - 1)
        
    values[n] = resultat
    return resultat
    
    
    

temps_1 = timeit(
    stmt=lambda : factoriel(80),
    number=5000
)

temps_2 = timeit(
    stmt=lambda : facto(80),
    number=5000
)

temps_3 = timeit(
    stmt=lambda : facto_recursif(80),
    number=5000
)

print(temps_1)
print(temps_2)
print(temps_3)
```
