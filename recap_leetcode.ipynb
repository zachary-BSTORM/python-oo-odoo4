```python
value = "7"
total = 0

if "0" <= value <= "9": # chiffre ASCII uniquement
    total += ord(value) - ord("0") # on soustrait ( 55 - 48) pour récupérer la valeur
```

formattage de minutes

```python
heures = 6
minutes = 5

# version manuel
minutes = ("0" + str(minutes)) if minutes < 10 else minutes

# formattage 
f"{heures}h{minutes:02d}"

print(minutes)
```

première occurence d'un caractère

```python
# mauvais
count(5) # pour chaque caractère sera plus lourd (O(n²))

# mieux
# parcourir et ajouter dans un dict: O(n)
s = "abcdefabcde"
def premier_unique(s):
    values = {}
    for v in s:
        values[v] = values.get(v,0) + 1
        
    # parcourir le dictionnaire une seule fois
    for v in s:
        if values[v] == 1:
            return v
    return "aucun"
    
```
