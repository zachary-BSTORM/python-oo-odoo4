première occurence d'un caractère

```python
# mauvais
count(5) # pour chaque caractère sera plus lourd (O(n²))

# mieux
# parcourir et ajouter dans un dict: O(n)
s = "abcdefabcde"
def premier_unique(s):
    values = {}
    for v in s:
        values[v] = values.get(v,0) + 1
        
    # parcourir le dictionnaire une seule fois
    for v in s:
        if values[v] == 1:
            return v
    return "aucun"
    
```

verification différence entre plusieurs valeurs

```python
# version pythonique
ecart = nombres[1] - nombres[0]
if all(nombres[i] - nombres[i - 1] == ecart for i in range(2, len(nombres))):
    return "arithmetique"


# version normale
ecart = nombres[1] - nombres[0]
est_arithmetique = True
for i in range(2, len(nombres)):
    if nombres[i] - nombres[i - 1] != ecart:
        est_arithmetique = False
        break                       # inutile de continuer
if est_arithmetique:
    return "arithmetique"

# Solution 4
def type_suite(nombres):
    
    ecart = nombres[1] - nombres[0] # récupération du premier écart
    if all(nombres[i] - nombres[i - 1] == ecart for i in range(2, len(nombres))): # vérification
        return "arithmetique"
    
    # rapport constant : n[i] / n[i-1] == n[1] / n[0], écrit sans division
    a, b = nombres[0], nombres[1] # récupération des valeurs pour la comparaison
    if all(nombres[i] * a == nombres[i - 1] * b for i in range(2, len(nombres))): # vérification
        return "geometrique"
    
    return "aucune" # valeur par défaut
```

Correction exo 5

```python
def parentheses_equilibrees(s):
    paires = {")": "(", "]": "[", "}": "{"}
    pile = [] # une liste utilisée comme pile
    for c in s:
        if c in "([{":
            pile.append(c) # on empile chaque ouvrant
        elif c in paires:
            if not pile or pile.pop() != paires[c]:
                return "false" # rien à fermer, ou mauvais ouvrant
    return "true" if not pile else "false" # tout doit être refermé
```

Exercice 1 Premier client récurrent

```python
def premier_client_recurrent(commandes, minimum):
compteur = {} # nom normalisé -> nombre de commandes
premier_nom = {} # nom normalisé -> nom tel qu'écrit la 1re fois
for commande in commandes:
nom = commande["client"].strip()
cle = nom.lower()
if cle not in premier_nom:
premier_nom[cle] = nom
compteur[cle] = compteur.get(cle, 0) + 1
if compteur[cle] == minimum:
return premier_nom[cle]
return "-1"
```

déplacement de zero

```python
def deplacer_zeros(nombres):
    ecriture = 0 # prochaine place pour une valeur non nulle
    for lecture in range(len(nombres)):
        if nombres[lecture] != 0:
            nombres[ecriture], nombres[lecture] = nombres[lecture], nombres[ecriture]
            ecriture += 1
    return nombres
```

Exercice 3 Même motif 

```python
def meme_motif(a, b):
    if len(a) != len(b):
        return "false"
    a_vers_b, b_vers_a = {}, {}
    for x, y in zip(a, b):
        if a_vers_b.setdefault(x, y) != y or b_vers_a.setdefault(y, x) != x:
            return "false"
    return "true"

print(meme_motif("paperz", "aaaaaa"))
```

La liste nombres est triée par ordre croissant et peut contenir des doublons.
Écrire une fonction qui renvoie [premier_indice, dernier_indice] de cible, ou [-1, -1] si cible est
absente.

```python
def premiere_derniere(nombres, cible):
    def borne(premiere):
        g, d, res = 0, len(nombres) - 1, -1
        while g <= d:
            m = (g + d) // 2
            if nombres[m] == cible:
                res = m  # on note, puis on continue
                if premiere:
                    d = m - 1
                else:
                    g = m + 1
            elif nombres[m] < cible:
                g = m + 1
            else:
                d = m - 1
        return res

    return [borne(True), borne(False)]


print(premiere_derniere([1, 2, 2, 2, 5, 7], 2))
```

```python
def fenetres_distinctes(texte, k):
    if k > len(texte):
        return 0
    compte, resultat = {}, 0
    for i, c in enumerate(texte):
        compte[c] = compte.get(c, 0) + 1 # caractère entrant
        if i >= k:
            sortant = texte[i - k] # caractère sortant
            compte[sortant] -= 1
        if compte[sortant] == 0:
            del compte[sortant]
        if i >= k - 1 and len(compte) == k:
            resultat += 1
    return resultat
```
